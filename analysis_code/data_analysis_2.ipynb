### Adult Income Prediction Dataset

**Aim**: To apply explatory data analysis to uncover important insights from the dataset of individuals from the 1994 U.S. Census database.

**Source**: The *Adult Income Prediction Dataset* dataset was obtained from [Kaggle](https://www.kaggle.com/datasets/mosapabdelghany/adult-income-prediction-dataset?select=adult.csv) website

### 1. Import libraries

```python
# import required python libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# histogram plot for distribution of age

# extract the age column of the dataset
age_col = adult_df['age']

# plot the histogram
plt.hist(x=age_col, bins=20)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.show()
```

```python
# distribution of adults by education level

# extract the counts for each education level
edu_col = adult_df['education'].value_counts()

# specify the plot size 
plt.figure(figsize=(20, 10))

# plot the horizontal bar chart 
bar_chrt_edu = plt.barh(edu_col.index, edu_col.values)
plt.title('Distribution of Education Level')
plt.ylabel('Education Level')
plt.xlabel('Count')

# add labels on top of the bars
plt.bar_label(bar_chrt_edu, fmt='%.0f')

plt.show()
```

```python
# distribution of adults by work class

# extract the counts for each work class
wrkclass_col = adult_df['workclass'].value_counts()

# specify the plot size 
plt.figure(figsize=(20, 10))

# plot the horizontal bar chart 
bar_chrt_wrk = plt.barh(y=wrkclass_col.index, width=wrkclass_col.values)
plt.title('Distribution of Work Class')
plt.xlabel('Count')
plt.ylabel('Work Class')

# add labels on top of the bars
plt.bar_label(bar_chrt_wrk, fmt='%.0f')

plt.show()
```

```python
# distribution of adults by gender

# extract the counts for each gender
gender_col = adult_df['sex'].value_counts()

# plot the horizontal bar chart
plt.barh(y=gender_col.index, width=gender_col.values)
plt.title('Distribution of Gender')
plt.xlabel('Count')
plt.ylabel('Gender')

plt.show()
```

```python
# distribution of income across the adult population

# extract the counts for each salary range
inc_dist = adult_df['income'].value_counts()

# extract the income categories and corresponding counts
x, y = inc_dist.index, inc_dist.values

# plot the bar chart
plt.bar(x, y)
plt.title('Distribution of Income')
plt.xlabel('Income')
plt.ylabel('Count')

plt.show()

```

### 4. Cleaning the Dataset

```python
# check for null values
adult_df.isna().sum()
```

```python
# remove the occurence of "?"" from the dataset

rows_to_maintain = adult_df.astype(str).apply(lambda x: x.str.contains(r'\?').any(), axis=1)

new_adult_income_df = adult_df[~rows_to_maintain]

# view the first few rows of the clean dataset
new_adult_income_df.head(10)
```

```python
# drop the current index of the clean dataset
new_adult_income_df.reset_index(inplace=True, drop=True)

# view the first few rows of the clean dataset
new_adult_income_df.head(10)
```

```python
# check the size of the clean dataset
new_adult_income_df.shape
```

```python
# check the size of the original dataset
adult_df.shape
```

```python
# determine the percentage of data left after cleaning

# no of records in original dataset
len_raw = len(adult_df)

# no of records in clean dataset
len_clean = len(new_adult_income_df)

# calculate the percentage of data left
percent_rem_data = (len_clean / len_raw) * 100

percent_rem_data
```

### 5. Feature Relationship

```python
# correlation between the columns 
corr_df_values = new_adult_income_df[['age', 'fnlwgt', 'education.num',
       'capital.gain', 'capital.loss', 'hours.per.week']].corr()

# use seaborn to visualize the correlation
sns.heatmap(corr_df_values, annot=True)
plt.show()
```

### 6. Asking Questions

#### 1. How does hours worked per week influence the likelihood of earning over or less than $50K per year?

```python
# group the hours worked per week by income earned
inc_by_hrs_wrk = new_adult_income_df.groupby('hours.per.week')[['income']].value_counts()

# convert the Series object into a Dataframe
plot_df = inc_by_hrs_wrk.unstack(fill_value=0)

# specify the plot figure size
plt.figure(figsize=(20, 12))

# plot the first stack
bar1 = plt.bar(plot_df.index, plot_df['<=50K'], label='<=50K', color='green')

# plot the second stack, with its bottom positioned at the top of the first stack
bar2 = plt.bar(plot_df.index, plot_df['>50K'], bottom=plot_df['<=50K'], label='>50K', color='orange')

# get the current Axes object for axis manipulation
ax = plt.gca()

# set the title, x-axis, y-axis and legend
plt.title('Distribution of Income by Hours Worked per Week')
plt.xlabel('Hours per week')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()

plt.show()
```

#### 2. How does education level impact the likelihood of earning over or less than $50K per year?

```python

# group the education level by income earned
inc_by_edu_lvl = new_adult_income_df.groupby('education')[['income']].value_counts()

# convert the Series object into a Dataframe
plot_df = inc_by_edu_lvl.unstack(fill_value=0)


# specify the plot figure size
plt.figure(figsize=(10, 6))

# plot the first stack
bar1 = plt.bar(plot_df.index, plot_df['<=50K'], label='<=50K', color='green')

# plot the second stack, with its bottom positioned at the top of the first stack
bar2 = plt.bar(plot_df.index, plot_df['>50K'], bottom=plot_df['<=50K'], label='>50K', color='blue')


# get the current Axes object for axis manipulation
ax = plt.gca()

# set the background of the plot
ax.set_facecolor((0.8, 0.8, 0.8, 1))

# add labels and remove the y-axis
ax.bar_label(bar1, label_type='center', color='white', fontsize=10, fontweight='bold')
ax.bar_label(bar2, label_type='center', color='white', fontsize=10, fontweight='bold')
ax.get_yaxis().set_visible(False)

# remove the top, right, and left borders for a cleaner look
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)

# set the title, x-axis, y-axis and legend
plt.title('Distribution of Income by Education Level')
plt.xlabel('Education Level')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()

plt.show()
```

#### 3. Is there a significant income gap between genders?

```python
# group the gender by income earned
inc_by_sex = new_adult_income_df.groupby('sex')[['income']].value_counts()

# convert the Series object into a Dataframe
plot_df = inc_by_sex.unstack(fill_value=0)

# specify the plot figure size
plt.figure(figsize=(10, 6))

# plot the first stack
bar1 = plt.bar(plot_df.index, plot_df['<=50K'], label='<=50K', color='orange')

# plot the second stack, with its bottom positioned at the top of the first stack
bar2 = plt.bar(plot_df.index, plot_df['>50K'], bottom=plot_df['<=50K'], label='>50K', color='purple')


# get the current Axes object for axis manipulation
ax = plt.gca()


# add labels and remove the y-axis
ax.bar_label(bar1, label_type='center', color='white', fontsize=10, fontweight='bold')
ax.bar_label(bar2, label_type='center', color='white', fontsize=10, fontweight='bold')
ax.get_yaxis().set_visible(False)

# remove the top, right, and left borders for a cleaner look
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)

# set the title, x-axis, y-axis and legend
plt.title('Distribution of Income by Gender')
plt.xlabel('Gender')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()

plt.show()
```

#### 4.  Which occupations have the highest earning potential?

```python
# group the education level by income earned
inc_by_occu = new_adult_income_df.groupby('occupation')[['income']].value_counts()

# convert the Series object into a Dataframe
plot_df = inc_by_occu.unstack(fill_value=0)

# specify the plot figure size
plt.figure(figsize=(10, 6))

# plot the first stack
bar1 = plt.bar(plot_df.index, plot_df['<=50K'], label='<=50K', color='brown')

# plot the second stack, with its bottom positioned at the top of the first stack
bar2 = plt.bar(plot_df.index, plot_df['>50K'], bottom=plot_df['<=50K'], label='>50K', color='black')


# get the current Axes object for axis manipulation
ax = plt.gca()

# set the background of the plot
ax.set_facecolor((0.8, 0.8, 0.8, 1))

# add labels and remove the y-axis
ax.bar_label(bar1, label_type='center', color='white', fontsize=10, fontweight='bold')
ax.bar_label(bar2, label_type='center', color='white', fontsize=10, fontweight='bold')
ax.get_yaxis().set_visible(False)

# remove the top, right, and left borders for a cleaner look
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)

# set the title, x-axis, y-axis and legend
plt.title('Distribution of Income by Occupation')
plt.xlabel('Occupation')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()

plt.show()
```

#### 5. How does marital status and relationship role influence income?

```python
# group the marital status by income earned
inc_by_mar_status = new_adult_income_df.groupby('marital.status')[['income']].value_counts()

# group the relationship status by income earned
inc_by_rel = new_adult_income_df.groupby('relationship')[['income']].value_counts()

# convert the two Series object into Dataframes
plot_df_1 = inc_by_mar_status.unstack(fill_value=0)
plot_df_2 = inc_by_rel.unstack(fill_value=0)

# specifying the subplots parameters
no_of_subplots = 2
rows = 1
cols = 2

# specify the plot figure size
plt.figure(figsize=(15, 10))

# subplot 1: for the distribution of income by marital status
plt.subplot(rows, cols, 1)
# plot the first stack
bar1 = plt.bar(plot_df_1.index, plot_df_1['<=50K'], label='<=50K', color='#A020F0')
# plot the second stack, with its bottom positioned at the top of the first stack
bar2 = plt.bar(plot_df_1.index, plot_df_1['>50K'], bottom=plot_df_1['<=50K'], label='>50K', color='#B22222')
# get the current Axes object for axis manipulation
ax = plt.gca()
# set the background of the plot
ax.set_facecolor((0.8, 0.8, 0.8, 1))
# add labels and remove the y-axis
ax.bar_label(bar1, label_type='center', color='white', fontsize=10, fontweight='bold')
ax.bar_label(bar2, label_type='center', color='white', fontsize=10, fontweight='bold')
ax.get_yaxis().set_visible(False)
# remove the top, right, and left borders for a cleaner look
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
# set the title, x-axis, y-axis and legend
plt.title('Distribution of Income by Marital Status')
plt.xlabel('Marital Status')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()



# subplot 2: for the distribution of income by relationship status
plt.subplot(rows, cols, 2)
# plot the first stack
bar3 = plt.bar(plot_df_2.index, plot_df_2['<=50K'], label='<=50K', color='#228B22')
# plot the second stack, with its bottom positioned at the top of the first stack
bar4 = plt.bar(plot_df_2.index, plot_df_2['>50K'], bottom=plot_df_2['<=50K'], label='>50K', color='#004D99')
# get the current Axes object for axis manipulation
ax = plt.gca()
# set the background of the plot
ax.set_facecolor((0.8, 0.8, 0.8, 1))
# add labels and remove the y-axis
ax.bar_label(bar3, label_type='center', color='white', fontsize=10, fontweight='bold')
ax.bar_label(bar4, label_type='center', color='white', fontsize=10, fontweight='bold')
ax.get_yaxis().set_visible(False)
# remove the top, right, and left borders for a cleaner look
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
# set the title, x-axis, y-axis and legend
plt.title('Distribution of Income by Relationship')
plt.xlabel('Relationship Status')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()


# add a title to the subplots and adjust the spacing between the title and subplots
plt.suptitle('Distribution of Income based on Marital Status and Relationship', fontsize=16)
plt.subplots_adjust(top=0.9)

plt.show()
```
